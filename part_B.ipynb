## Problem 5. Array operations

```python
import numpy as np

values = np.arange(1, 101, dtype=float)
print("Mean:", np.mean(values))
print("Median:", np.median(values))
print("Variance:", np.var(values))
print("Standard deviation:", np.std(values))
print("Minimum:", values.min(), "at index", values.argmin())
print("Maximum:", values.max(), "at index", values.argmax())

matrix = values.reshape(10, 10)
print("Matrix shape:", matrix.shape)
print("First row:", matrix[0])
print("Rows 2–4 and columns 3–5:\n", matrix[1:4, 2:5])
print("Values greater than 90:", values[values > 90])
```

## Problem 6. Matrix operations

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[6, 5, 4], [3, 2, 1]])

print("A + B:\n", A + B)
print("Element-wise multiplication A * B:\n", A * B)
print("Transpose of A:\n", A.T)
print("Matrix product A @ B.T:\n", A @ B.T)

print("Element-wise multiplication multiplies matching positions.")
print("Matrix multiplication combines rows and columns; the inner dimensions must match.")
```

## Problem 7. Normalization and standardization

Min-max normalization:

\[
x' = \frac{x - x_{\min}}{x_{\max} - x_{\min}}
\]

Z-score standardization:

\[
z = \frac{x - \mu}{\sigma}
\]

If all values are identical, the denominator is zero; this implementation returns zeros instead of dividing by zero.

```python
def min_max_normalize(values):
    values = np.asarray(values, dtype=float)
    low, high = values.min(), values.max()
    if low == high:
        return np.zeros_like(values)
    return (values - low) / (high - low)

def z_score_standardize(values):
    values = np.asarray(values, dtype=float)
    standard_deviation = values.std()
    if standard_deviation == 0:
        return np.zeros_like(values)
    return (values - values.mean()) / standard_deviation

sample = np.array([10, 20, 30, 40, 50], dtype=float)
identical = np.array([7, 7, 7], dtype=float)

print("Original:", sample)
print("Min-max normalized:", min_max_normalize(sample))
print("Z-score standardized:", z_score_standardize(sample))
print("Identical values, min-max:", min_max_normalize(identical))
print("Identical values, z-score:", z_score_standardize(identical))

print("Min-max normalization maps values to [0, 1] and is useful when a bounded range is wanted.")
print("Z-score standardization centres values around zero with unit standard deviation; it is useful for many scale-sensitive models.")
```
